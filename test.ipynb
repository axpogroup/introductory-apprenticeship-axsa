```python
import pandas as pd

df = pd.read_csv('/Users/finnratzinger/Documents/introductory-apprenticeship-axsa/web_scraper/clean_data.csv')

print(type(df))          # Was für ein Objekt ist df?
print(type(df['name']))  # Was für ein Objekt ist eine einzelne Spalte?

print(df.columns)        # Alle Spaltennamen anzeigen
print(df.shape)          # (Zeilen, Spalten)
```

```python
# Eine Spalte auswählen - gibt eine Series zurück
namen = df['name']
print(namen.head())

print()

# Mehrere Spalten auswählen - gibt einen DataFrame zurück (Achtung: doppelte Klammern!)
teil_tabelle = df[['name', 'price']]
print(teil_tabelle.head())
```
